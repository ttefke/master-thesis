```python
# Initial values
n = 10**6
N = 8
m = 9
M = n/N
```

```python
N <= 100
```

```python
M > (0.01 * n)
```

```python
N == int(n/M)
```

```python
M
```

```python
# Compute mean:
mean = (M-m+1)/(2**m) # (Length of substring - length of template + 1) /  2 ** length of substring
mean
```

```python
# Variance
var = M * ((1/(2**m)) - (2*m-1)/(2**(2*m)))
var
```

```python
std = var ** 0.5
std
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
x = np.arange(-1000, 1500, 1)
plt.plot(x, norm.pdf(x, mean, std), label="std")
plt.plot(x, norm.pdf(x, mean, var), label="var")
plt.legend()
plt.savefig("../figures/t7.pdf")
plt.show()
```

```python
w = np.arange(0, 500, 1)
y_var = ((w - mean)**2)/var
y_std = ((w - mean)**2)/std   
```

```python
plt.plot(w, y_std, label="std")
plt.plot(w, y_var, label="var")
plt.legend()
plt.show()
```

```python
from scipy import special
special.gammaincc(1, (2.1333/2))
```

```python
res_var = special.gammaincc(N/2, y_var/2)
res_std = special.gammaincc(N/2, y_std/2)
```

```python
plt.plot(w, res_std, label="std")
plt.plot(w, res_var, label="var")
plt.legend()
plt.show()
```
