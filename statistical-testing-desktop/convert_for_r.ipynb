```python
MIN = 0
MAX = 250
DEVICE = "chaoskey"
```

```python
import os
for i in range(MIN+1, MAX+1):
    original_file = f"./data/{DEVICE}/{i:03d}"
    txt = f"{original_file}.txt"
    os.rename(original_file, txt)
```

```python
# Convert to array

all_bytes = []
all_bytes_32 = []
for i in range(MIN+1, MAX+1):
    original_file = f"./data/{DEVICE}/{i:03d}"
    txt = f"{original_file}.txt"
    b_f = f"{original_file}.bin"
    csv_f = f"{original_file}.csv"
    bytes_ = []

    with open(txt, "r") as f:
        contents = f.readlines()
        for line in contents:
            line = line.strip()
            for i in range(0, len(line), 8):
                byte = line[i:i+8]
                all_bytes.append(int(byte, base=2))
                bytes_.append(int(byte, base=2))
                
            for i in range(0, len(line), 32):
                byte = line[i:i+32]
                all_bytes_32.append(int(byte, base=2))

    with open(b_f, "wb") as f:
        f.write(bytearray(bytes_))

    with open(csv_f, "w") as f:
        for byte_ in bytes_:
            f.write(f"{byte_}\n")
        
with open(f"./data/{DEVICE}/all.csv", "w") as f:
    for byte_ in all_bytes:
            f.write(f"{byte_}\n")
            
with open(f"./data/{DEVICE}/all_{DEVICE}_32.csv", "w") as f:
    for byte_ in all_bytes_32:
            f.write(f"{byte_}\n")
```
